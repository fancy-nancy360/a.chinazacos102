```python
import tkinter as tk
from tkinter import messagebox

# Parent class
class Zenith:
    def __init__(self, division_name):
        self.division_name = division_name

    def unique_services(self):
        """To be overridden by subclasses"""
        return []

    def mutual_services(self):
        """Common services across divisions"""
        return []

    def get_all_services(self):
        """Combines unique and mutual services"""
        return self.unique_services() + self.mutual_services()


# Retail Banking subclass
class RetailBanking(Zenith):
    def unique_services(self):
        return [
            "Retirement and education accounts",
            "Loans and mortgages",
            "Checking and saving"
        ]

    def mutual_services(self):
        return [
            "Lines of credit",
            "Investment management and accounts",
            "Insurance"
        ]


# Global Banking subclass
class GlobalBanking(Zenith):
    def unique_services(self):
        return [
            "Multi-currency management services and products",
            "Foreign currency accounts",
            "Foreign currency credit cards",
            "Transborder advisory services",
            "Liquidity management"
        ]

    def mutual_services(self):
        # Global Banking has no mutual services with others here
        return []


# Commercial Banking subclass
class CommercialBanking(Zenith):
    def unique_services(self):
        return [
            "Advisory services"
        ]

    def mutual_services(self):
        return [
            "Lines of credit",
            "Investment management and accounts",
            "Insurance"
        ]


# Factory function to get appropriate division object (polymorphism)
def get_division(division_name):
    if division_name == "Retail Banking":
        return RetailBanking(division_name)
    elif division_name == "Global Banking":
        return GlobalBanking(division_name)
    elif division_name == "Commercial Banking":
        return CommercialBanking(division_name)
    else:
        return None


# GUI Application
class ZenithApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Zenith Bank Service Manager")
        self.root.geometry("500x450")
        self.root.resizable(False, False)

        # Title label
        title = tk.Label(root, text="Zenith Bank Division Services", font=("Arial", 16, "bold"))
        title.pack(pady=10)

        # Employee Name
        tk.Label(root, text="Employee Name:", font=("Arial", 12)).pack(pady=(10, 0))
        self.name_entry = tk.Entry(root, font=("Arial", 12), width=30)
        self.name_entry.pack(pady=5)

        # Division selection
        tk.Label(root, text="Select Division:", font=("Arial", 12)).pack(pady=(10, 0))
        self.division_var = tk.StringVar(value="Retail Banking")
        divisions = ["Retail Banking", "Global Banking", "Commercial Banking"]
        self.division_menu = tk.OptionMenu(root, self.division_var, *divisions)
        self.division_menu.config(font=("Arial", 12), width=28)
        self.division_menu.pack(pady=5)

        # Submit button
        self.submit_btn = tk.Button(root, text="Show Services", font=("Arial", 12, "bold"), bg="blue", fg="white", command=self.show_services)
        self.submit_btn.pack(pady=15)

        # Result display area
        self.result_text = tk.Text(root, height=12, width=55, font=("Arial", 10))
        self.result_text.pack(pady=10)
        self.result_text.config(state=tk.DISABLED)

        # Footer
        footer = tk.Label(root, text="© Zenith Bank - OOP Demo", font=("Arial", 8))
        footer.pack(side=tk.BOTTOM, pady=5)

    def show_services(self):
        employee_name = self.name_entry.get().strip()
        division_name = self.division_var.get()

        if not employee_name:
            messagebox.showwarning("Input Error", "Please enter the employee name.")
            return

        # Get division object (polymorphism)
        division = get_division(division_name)
        if not division:
            messagebox.showerror("Error", "Invalid division selected.")
            return

        # Get services
        all_services = division.get_all_services()
        unique_services = division.unique_services()
        mutual_services = division.mutual_services()

        # Prepare output
        output = f"Employee: {employee_name}\n"
        output += f"Division: {division.division_name}\n"
        output += "-" * 40 + "\n"
        output += "SERVICES RENDERED:\n\n"

        if unique_services:
            output += "🔹 Unique Services:\n"
            for service in unique_services:
                output += f"   • {service}\n"
            output += "\n"

        if mutual_services:
            output += "🔸 Mutual Services:\n"
            for service in mutual_services:
                output += f"   • {service}\n"
            output += "\n"

        if not all_services:
            output += "No services available.\n"

        # Display in text area
        self.result_text.config(state=tk.NORMAL)
        self.result_text.delete(1.0, tk.END)
        self.result_text.insert(tk.END, output)
        self.result_text.config(state=tk.DISABLED)


# Run the application
if __name__ == "__main__":
    root = tk.Tk()
    app = ZenithApp(root)
    root.mainloop()
```
