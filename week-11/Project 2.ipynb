```python

import tkinter as tk
from tkinter import messagebox as mesgbx
from tkinter import scrolledtext as scroltxt
from abc import ABC, abstractmethod 
import pandas as pd

class External_Vendors(ABC):
    @abstractmethod
    def vendor_check(self):
        pass

class Vendors(External_Vendors):

    Menu_items = []
    Prices = []

    def __init__ (self, file):
        self.data = pd.read_csv(file)

    def vendor_check(self):
        for index, row in self.data.iterrows():
            self.Menu_items.append(row["Meal"])
            self.Prices.append(row["Price"])

class GUI(Vendors):
    def __init__(self):
        self.root = tk.Tk()
        self.root.geometry("700x500")
        self.root.title("PAU Vendor Management System.")

        self.label1 = tk.Label(self.root, text = "Welcome to PAU's Vendor Menu Management.", font = ("Helevitica", 20), fg = "blue")
        self.label1.pack(pady = 20)
        self.label2 = tk.Label(self.root, text = "Which Vendor are you?\n\n1. Cooperative Cafe\n2. Faith Hostel Cafe\n3. Student Center Cafe", font = ("Arial",14))
        self.label2.pack(pady = 10)

        self.vendor_entry = tk.Entry(self.root)
        self.vendor_entry.pack(pady = 5)

        self.displaybutton = tk.Button(self.root, text = "Display menu", command = self.menu, font = ("Arial", 14), bg = "blue", fg = "white")
        self.displaybutton.pack(pady = 10)

        self.clearbutton = tk.Button(self.root, text = "Clear all", command = self.clear_all, font = ("Arial", 14), bg = "red", fg = "white")
        self.clearbutton.pack(pady = 5)

        self.result_window = scroltxt.ScrolledText(self.root, width = 50, height = 15)
        self.result_window.pack(pady = 5)
    
    def menu(self):
        vendor_name = self.vendor_entry.get().lower()
        if vendor_name == 'cooperative cafe':
            vendor = Vendors("Cooperative cafe.csv")
            vendor.vendor_check()
            self.placement_in_scrolltext()
        elif vendor_name == 'faith hostel cafe':
            vendor = Vendors("Faith hostel cafe.csv")
            vendor.vendor_check()
            self.placement_in_scrolltext()
        elif vendor_name == 'student center cafe':
            vendor = Vendors("Student center cafe.csv")
            vendor.vendor_check()
            self.placement_in_scrolltext()
        else:
            mesgbx.showerror("Error", "Your vendor was not found\nPlease choose one of the vendors shown.")

    def placement_in_scrolltext(self):
        vendor_name = self.vendor_entry.get().title()
        self.result_window.insert(tk.END, f"\n{vendor_name} Menu\n{"=" * (5 + len(vendor_name))}")
        for index in range(0,len(self.Menu_items)):
            self.result_window.insert(tk.END, f"\n{self.Menu_items[index]} N{self.Prices[index]}\n")
        self.Menu_items.clear()
        self.Prices.clear()

    def clear_all(self):
        self.vendor_entry.delete(0, tk.END)
        self.result_window.delete(1.0, tk.END)

    def run(self):
        self.root.mainloop()

Vendor_app = GUI()
Vendor_app.run()
```
