```python
from abc import ABC, abstractmethod

class coup_de_ecriva(ABC):
    @abstractmethod
    def Fan_page(self):
        pass

class Man_United(coup_de_ecriva):
    def Fan_page(self):
        return "Welcome to Man_United"
    
class Real_Madrid(coup_de_ecriva):
    def Fan_page(self):
        return "Welcome to Real_Madrid"
    
class Barcelona(coup_de_ecriva):
    def Fan_page(self):
        return "Welcome to Barcelona"
    
class PSG(coup_de_ecriva):
    def Fan_page(self):
        return "Welcome to PSG"
    
print("Welcome!\nLet's get you settled into your club.")
club_choice = input("Which club are you a fan of?\n1. Man_United\n2. Real_Madrid\n3. Barcelone\n4. PSG\nEnter the number corresponding to your choice: ")
if club_choice == "1":
    club = Man_United()
    print(club.Fan_page())
elif club_choice == "2":
    club = Real_Madrid()
    print(club.Fan_page())
elif club_choice == "3":
    club = Barcelona()
    print(club.Fan_page())
elif club_choice == "4":
    club = PSG()
    print(club.Fan_page())
else:
    print("Invalid choice. Please enter a number corresponding to your choice.")
```
