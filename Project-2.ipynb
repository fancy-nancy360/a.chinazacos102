```python
#to PAU, >=10: 2000, <10: 1500.
#to Epe, >=10: 5000, <10: 4000.
import tkinter as tk
from tkinter import messagebox as mesgbx

def weight_input():
    while True:
        try:
            weight = float(weight_entry.get())
            return weight
        except ValueError:
            mesgbx.showerror("Invalid weight", "Couldn't process the input for weight.")
            return None


def show_price(price):
    mesgbx.showinfo("Payment successfully processed!",f"Your bill is ₦{price}")
    payment_label.config(text = f"Your bill is ₦{price}.00")
    payment_label.pack()

class product:
    def __init__(self, weight, destination):
        self.weight = weight
        self.destination = destination

    def payment(self):
        if self.weight <= 0:
            mesgbx.showerror("Invalid weight", "Weight must be a positive number.")
        elif self.weight >= 10 and self.destination.upper() == "PAU":
            show_price(2000)
        elif self.weight <= 10 and self.destination.upper() == "PAU":
            show_price(1500)
        elif self.weight <= 10 and self.destination.lower() == "epe":
            show_price(4000)
        elif self.weight >= 10 and self.destination.lower() == "epe":
            show_price(5000)
        else:
            show_price(3000)

def process_payment():
    """Called on button press — reads inputs fresh, validates, then runs payment."""
    weight = weight_input()           
    if weight is None:
        return                      

    destination = destination_entry.get().strip()
    if not destination:
        mesgbx.showerror("Invalid destination", "Please enter a destination.")
        return

    product_ordered = product(weight, destination)
    product_ordered.payment()

root = tk.Tk()
root.geometry("800x800")
root.title("Product delivery system.")

label1 = tk.Label(root, text = "Welcome to Jose's product delivery application.\nWhat would you like to ship.")
label1.pack()

weight_label = tk.Label(root, text = "Weight of product.(Kg)")
weight_label.pack()
weight_entry = tk.Entry(root)
weight_entry.pack()

destination_label = tk.Label(root, text = "Destination of product.")
destination_label.pack()
destination_entry = tk.Entry(root)
destination_entry.pack()

process_button = tk.Button(root, text = "process payment", command = process_payment)
process_button.pack()

payment_label = tk.Label(root, text = "")

root.mainloop()
```
