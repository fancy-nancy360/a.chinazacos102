```python
import tkinter as tk
from tkinter import messagebox as msgbox
import random

employees = ["Mary Evans", "Eyo Ishan", "Durojaiye Dare", "Adams Ali", "Andrew Ugwu", "Stella Mankinde", "Jane Akibo", "Ago James", "Michell Taiwo", "Abraham Jones" , "Nicole Anide", "Kosi Korso", "Adele Martins", "Emmanuel Ojo", "Ajayi Fatima"]
tasks = ["Loading", "Transporting", "Reveiwing Orders", "Customer Service", "Delivering Items"]

def task_assesment():            
    if task_status.get():
        msgbox.showinfo("Task completed!","Your completed task has been logged.\nGood job.")
        root.destroy()
    else:
        msgbox.showerror("Task Incomplete!","Sorry, the task has not been confirmed for completion.")


class employee:
    name = ""
    attendance = False

    def assign_task(self):
        global task, task_status
        task_id = random.randint(0,4)
        task = tasks[task_id]
        msgbox.showinfo("Assigned task",f"Your task: {task}.\nWhen you have completed it, Click the check box in the window")
        task_status = tk.IntVar(value = 0)
        task_complete_check_button = tk.Checkbutton(root, text = "Completed task?", variable = task_status)
        task_complete_check_button.pack()
        task_confirmation_button = tk.Button(root, text = "Confirm task", command = task_assesment)
        task_confirmation_button.pack()

    def check_employee(self):
        self.name = name_entry.get().title()
        if self.name not in employees:
            msgbox.showerror("Invalid","You don't seem to ba an employee here\nAccess to the system has been denied.")
            name_entry.delete(0,tk.END)
        else:
            msgbox.showinfo("Welcome.",f"Welcome back {self.name}")
            self.attendance = True
            self.assign_task()

root = tk.Tk()
root.title("Jane logistics")
root.geometry("500x500")

explanation_label = tk.Label(root, text = "Welcome to Jane's employee verification system.\nKindly enter your name.")
explanation_label.pack()

name_entry = tk.Entry(root)
name_entry.pack()

employee_to_log = employee()

submit_button = tk.Button(root, text = "Verify", command = employee_to_log.check_employee)
submit_button.pack()

root.mainloop()

```
