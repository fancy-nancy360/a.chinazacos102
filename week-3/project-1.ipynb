```python
import tkinter as tk
from tkinter import messagebox
import csv

employees = []

try:
    with open("GIG-logistics.csv", newline = '') as employee_file:
        reader = csv.reader(employee_file)
        header = next(reader)
        for record in reader:
            employees.append([field.strip() for field in record[1:4]])
except FileNotFoundError:
    print("Error. File not found.")
    exit()

def submit ():
    surname = surname_entry.get().strip().upper()
    firstname = firstname_entry.get().strip().title()
    department = department_entry.get().strip().title()
    if surname and firstname and department:
        if [surname, firstname, department] in employees:
            completed_authentication(surname, firstname, department)
        else:
            messagebox.showerror("Failed login","Invalid username or department")
    else:
            messagebox.showerror("Failed login","Please enter all parameters.")

def completed_authentication(surname, firstname, department):
    messagebox.showinfo(f"Welcome {firstname} {surname}\n Here are the others in your {department} department.")
    name = surname + ' ' + firstname
    collegue_list = [f"{n} {f}" for n, f, d in employees if d == department and n != name]
    collegue = "\n".join(collegue_list)

    messagebox.showinfo("Login sucscessful",f"Your collegues in {department} department are:\n {collegue}")

def clear():
    surname_entry.delete(0, tk.END)
    firstname_entry.delete(0, tk.END)
    department_entry.delete(0, tk.END)

root = tk.Tk()
root.title("GIG logistics authentication.")
root.geometry("700x350")

surname_label = tk.Label(root, text = "Enter your surname.")
surname_label.pack()
surname_entry = tk.Entry(root)
surname_entry.pack()

firstname_label = tk.Label(root, text = "Enter your firstname.")
firstname_label.pack()
firstname_entry = tk.Entry(root)
firstname_entry.pack()

department_label = tk.Label(root, text = "Enter your department.")
department_label.pack()
department_entry = tk.Entry(root)
department_entry.pack()

submition = tk.Button(root, text = "Process", command = submit)
submition.pack()

clear_button = tk.Button(root, text = "Clear", command = clear)
clear_button.pack()

root.mainloop()
```
